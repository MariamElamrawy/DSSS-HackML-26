# 07 · Submission
Retrain the chosen model (gradient boosting, 94 trees) on all of the training data and write `submission.csv`. `src/make_submission.py` does the same thing.

```python
import pandas as pd
from sklearn.ensemble import GradientBoostingClassifier

# Load ALL your clean data (no split this time!)
X_train = pd.read_csv("../Data/processed/X_train_clean.csv")
y_train = pd.read_csv("../Data/processed/y_train.csv")['overqualified']
X_test = pd.read_csv("../Data/processed/X_test_clean.csv")

# Load original test to get IDs
test_original = pd.read_csv("../Data/test.csv")

# Train on EVERYTHING
print("Training final model on all training data...")
final_model = GradientBoostingClassifier(n_estimators=94, random_state=42)
final_model.fit(X_train, y_train)

# Predict on test set
print("Making predictions on test set...")
test_predictions = final_model.predict_proba(X_test)[:, 1]

# Create submission file
submission = pd.DataFrame({
    'id': test_original['id'],
    'overqualified': test_predictions
})

# Save it
submission.to_csv("submission.csv", index=False)

print("\n✅ Submission file created!")
print(f"File: submission.csv")
print(f"Shape: {submission.shape}")
print("\nFirst few predictions:")
print(submission.head())
```
